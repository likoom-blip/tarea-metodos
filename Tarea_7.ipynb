```python
import numpy as np

def gauss_quad_5(f, a, b):
    def x(t):
        return ((b - a)*t + a + b)/2
    c = [0.2369268850, 0.4786286705, 0.5688888889, 0.4786286705, 0.2369268850]
    r = [0.9061798459, 0.5384693101, 0.0000000000, -0.5384693101, -0.9061798459]
    return ((b-a)/2)*sum(c[i]*f(x(r[i])) for i in range(5))

def f(x):
    return np.log(x)/np.sqrt(x + 5*x**3)

def g1(x):
    return 1/np.sqrt(1 + 5*x**2)
def g2(x):
    return 1/np.sqrt(5 + x**2)

a1 = [1, 0, -5/2, 0, 75/8]
a2 = [1/np.sqrt(5), 0, -1/(10*np.sqrt(5)), 0, 3/(200*np.sqrt(5))]

def tramo_singular(g, a, c):
    # integral de 0 a c de log(x) g(x) / sqrt(x)
    def resto(x):
        if x == 0:
            return 0.0
        P = sum(a[n]*x**n for n in range(len(a)))
        return np.log(x)*(g(x) - P)/np.sqrt(x)
    numerica = gauss_quad_5(resto, 0, c)
    analitica = sum(a[n]*c**(n + 0.5)*(np.log(c)/(n + 0.5) - 1/(n + 0.5)**2)
                    for n in range(len(a)))
    return numerica + analitica

A = tramo_singular(g1, a1, 0.2)

xs = np.linspace(0.2, 1, 9)
B = sum(gauss_quad_5(f, xs[i], xs[i+1]) for i in range(8))

C = -tramo_singular(g2, a2, 1.0)        # el signo menos viene del -log y

print(A, B, C, A + B + C)
```
