```python
#Tarea 4.8

import numpy as np

valores = np.array([1, -1]) #paso derecha e izquierda
probs = np.array([0.5, 0.5]) #probabilidad

mu = np.sum(valores*probs)
sigma2= np.sum(valores**2*probs) - mu**2

print("mu:", mu)
print("sigma^2:", sigma2)

N = 1000
trials = 10000

X = np.zeros(trials)
for i in range(trials):
  pasos = np.random.choice(valores, size=N, p=probs)
  X[i] = pasos.sum()

print("sigma simulada:", X.std(),  " vs teórica sqrt(N)*sigma:", np.sqrt(N * sigma2))
```
