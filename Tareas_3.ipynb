```python
#Tarea 3.1
import itertools

def unir(color, numeros):
    return {color + numero for numero in numeros}

urna = unir('B', '12345678') | unir('A', '123456') | unir('R', '123456789') #define las bolitas en la urna

def sin_roja_primeras_tres(e):
    return all(bola[0] != 'R' for bola in e[:3]) #checkea si en las primeras tres extraidas de la tupla 'e' son rojas o no, solo da True si no hay rojas.

def dos_de_cada_color(e):
    colores = [bola[0] for bola in e]
    return colores.count('B') == 2 and colores.count('A') == 2 and colores.count('R') == 2 #toma la primera letra de cada tupla en 'e' y luego ve cuantas 'B' hay, solo pasa el check si hay 2 de cada una.


total = 0
cuenta_a = 0
cuenta_b = 0

for e in itertools.permutations(urna, 6):
    total += 1
    if sin_roja_primeras_tres(e):
        cuenta_a += 1
    if dos_de_cada_color(e):
        cuenta_b += 1

print("a) P(ninguna roja entre las 3 primeras) =", cuenta_a / total)
print("b) P(2 de cada color) =", cuenta_b / total)
```
