```python
#Tarea 2.4.1
import numpy as np
#defino el termino a sumar y sus valores
N = 100

n_valores = np.arange(1, N +1, dtype=np.float16)

termino = np.float16(1.0) / (n_valores**np.float16(2))


suma = np.float16(0.0)
for i in termino:
  suma = np.float16(suma + i) #suma termino evaluado en 1 luego lo suma con el siguiente y así.

print (f"El resultado de la suma es : {suma}")
```

```python
#Tarea 2.4.2
import numpy as np
#defino el termino a sumar y sus valores
N = 100

n_valores = np.arange(1, N+1, dtype = np.float16)
termino = np.float16(1.0) / (n_valores**np.float16(2))

termino_alreves = termino [::-1] #parte del ultimo termino hasta el primero
#defino ciclo for para sumar termino a termino.
suma = np.float16(0.0)
for i in termino_alreves:
  suma = np.float16(suma + i)

print(f"El resultado de la suma es : {suma}")
```

```python
#Tarea 2.4.4
import numpy as np
N = 100
n_valores = np.arange(1, N+1, dtype = np.float16)
termino = np.float16(1.0) / (n_valores**np.float16(3))

termino_alreves = termino [::-1]
suma = np.float16(0.0)
for i in termino_alreves:
  suma = np.float16(suma + i)

print(f"El resultado de la suma : {suma}")
```

Al sumar en orden creciente, el acumulador crece rapido y los terminos pequeños terminan teniendo menor precisión y "se pierden". En orden decreciente, al empezar por los pequeños no se pierde información al acumularse con los grandes. Es decir, al sumar en orden creciente se tiene menos error.
